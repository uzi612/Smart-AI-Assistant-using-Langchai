```python
from pydantic import BaseModel

class PatientData(BaseModel):           # Step 1
    name : str
    age : int

def add_patient_data(patient : PatientData):
    print(patient.name, patient.age)
    print("Data added successfully !")

patient_data = {"name" : "AB", "age" : 25}
patient1 = PatientData(**patient_data)  # Step 2

add_patient_data(patient1)              # Step 3
```

```python
from pydantic import BaseModel
from typing import List, Dict, Optional

class PatientData(BaseModel):           # Step 1
    name : str
    age : int
    weight : float
    married : bool
    allergies : Optional[List[str]] = None
    contact_info : Dict[str, str]

def add_patient_data(patient : PatientData):
    print(patient.name, patient.age)
    print(patient.contact_info)
    print(patient.allergies)
    print("Data added successfully !")

patient_data = {"name" : "AB", "age" : 25,          # Preparing raw data
                "weight" : 70.1, "married" : False, 
                "allergies" : ['peanut', 'dust'], 
                "contact_info" : {"phone" : "9080821021", "email" : "abc@example.co"}}

patient1 = PatientData(**patient_data)  # Step 2

add_patient_data(patient1)              # Step 3
```

```python
from pydantic import BaseModel, EmailStr, Field
from typing import Optional

class PatientData(BaseModel):           # Step 1
    name : str = Field(min_length=3, max_length=15)
    age : int = Field(gt=0, lt=110)
    email : Optional[EmailStr] = None

def add_patient_data(patient : PatientData):
    print(patient.name, patient.age)
    print("Data added successfully !")

patient_data = {"name" : "AB", "age" : 23}          # Preparing raw data

patient1 = PatientData(**patient_data)  # Step 2

add_patient_data(patient1)              # Step 3
```

```python
from pydantic import BaseModel, Field
from typing import Optional, Annotated

class PatientData(BaseModel):           # Step 1
    name : Annotated[str, Field(min_length=3, max_length=15, title='Name of Patient', description='Here name of patient will be greater than 3 char and less than 15 char', examples=['sam', 'john'])]
    age : Annotated[int, Field(gt=0, lt=110)]
    email : Annotated[Optional[EmailStr], Field(default=None, title='Patients Email optional')]

def add_patient_data(patient : PatientData):
    print(patient.name, patient.age)
    print("Data added successfully !")

patient_data = {"name" : "AB", "age" : 23}          # Preparing raw data

patient1 = PatientData(**patient_data)  # Step 2

add_patient_data(patient1)              # Step 3
```

Custom Validators

```python
from pydantic import BaseModel, Field, field_validator
from typing import Optional, Annotated

class PatientData(BaseModel):           # Step 1
    name : str
    age : int
    email : EmailStr

    @field_validator['email']
    @classmethod
    def email_validator(cls, value):
        valid_domains = ['hdfc.com', 'icici.com']
        domain_name = value.split('@')[-1]
        if domain_name not in valid_domains:
            raise ValueError('Not a valid domain')

        return value

def add_patient_data(patient : PatientData):
    print(patient.name, patient.age)
    print("Data added successfully !")

patient_data = {"name" : "AB", "age" : 23}          # Preparing raw data

patient1 = PatientData(**patient_data)  # Step 2

add_patient_data(patient1)              # Step 3
```

Multiple Condition Validators

```python
from pydantic import BaseModel, model_validator

class PatientData(BaseModel):           # Step 1
    name : str
    age : int
    contact_details : Dict[str, str]

    @model_validator(mode='after')
    @classmethod
    def validate_emergency_contact(cls, model):
        if model.age > 69 and 'emergency' not in model.contact_details:
            raise ValueError('Patients older than 60 must have emergency contact')
        return model

def add_patient_data(patient : PatientData):
    print(patient.name, patient.age)
    print("Data added successfully !")

patient_data = {"name" : "AB", "age" : 23, "contact_datails":{'phone' : '9080989899'}}          # Preparing raw data

patient1 = PatientData(**patient_data)  # Step 2

add_patient_data(patient1)              # Step 3
```

Computed Fields

```python
from pydantic import BaseModel, computed_field
from typing import List, Dict, Optional

class PatientData(BaseModel):           # Step 1
    name : str
    age : int
    weight : float
    married : bool
    height : float

    @computed_field
    @property
    def bmi(self) -> float:
        bmi = round(self.weight/(self.height**2), 2)
        return bmi

def add_patient_data(patient : PatientData):
    print(patient.name, patient.age)
    print(patient.bmi)
    print("Data added successfully !")

patient_data = {"name" : "AB", "age" : 25,          # Preparing raw data
                "weight" : 70.1, "married" : False, "height" : 170.3}

patient1 = PatientData(**patient_data)  # Step 2

add_patient_data(patient1)              # Step 3
```
