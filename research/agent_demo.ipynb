```python
import os
import certifi
from dotenv import load_dotenv

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.tools.tavily_search import TavilySearchResults
from langchain.tools import tool
import requests
```

```python
from langchain_classic.agents import create_react_agent, AgentExecutor
```

```python
os.environ["SEL_CERT_FILE"] = certifi.where()
load_dotenv()

GEMINAI_KEY = os.getenv("GEMINAI_API_KEY")
TRAVILY_KEY = os.getenv("TRAVILY_API_KEY")
```

```python
from langsmith import Client

client = Client()
prompt = client.pull_prompt("hwchase17/react", dangerously_pull_public_prompt=True)
```

```python
# 1. Initialize the updated Tavily tool
search_tool = TavilySearchResults(max_results=1)
tools = [search_tool]

# 2. Use an active Gemini model
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")

# 3. Create the agent 
agent = create_react_agent(
    llm=llm,
    tools=tools,
    prompt=prompt
)
# 4. Create Executor
agent_executor = AgentExecutor(
    agent=agent,
    tools=tools,
    verbose=True
)

res = agent_executor.invoke({
    'input' : (
        "How is the a weather today in pune city"
    )
})
```
